```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('/content/drive/MyDrive/upi_transactions_2024.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
df.describe()
```

```python
df['amount'] = pd.to_numeric(df['amount (INR)'], errors='coerce')
df['fraud_flag'] = df['fraud_flag'].astype(int)
df['is_weekend'] = df['is_weekend'].astype(int)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
df.info()
df.describe()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
df.describe()
```

```python
df = df.rename(columns={
    'transaction_id': 'transaction_id',
    'amount_(inr)': 'amount'
})
df['timestamp'] = pd.to_datetime(df['timestamp'])
```

```python
df.info()
df.describe()
```

```python
df = df.loc[:, ~df.columns.duplicated()]
```

```python
df.info()
df.describe()
```

```python
df.to_csv("cleaned_upi_transactions.csv", index=False)
```
